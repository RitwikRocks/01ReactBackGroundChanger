# Crop Recommendation SYSTEM

## We will be using Classifaction Technique to predict the crop.


### Importing Libraries required for processing

```python
import pandas as pd
import numpy as np
```

### Importing the Data Set

```python
dataset = pd.read_csv("data.csv")
print(dataset['label'].unique())
X = dataset.iloc[:,:-1]
Y = dataset.iloc[:,-1].values
```

```python
print(X)
X['season'].unique()
```

### Encoding the categorical Data
We have to converting categorical data into numerical format.machine learning algorithms require numerical input to make predictions, these encoding methods simplify categorical variables, enabling algorithms to identify patterns and relationships in the data.

```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
X['season'] = le1.fit_transform(X['season'])
print(X['season'].unique())
```

```python
X = X.iloc[:, :].values
print(X)
```

```python
# from sklearn.compose import ColumnTransformer
# from sklearn.preprocessing import OneHotEncoder
# ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[1])],remainder='passthrough')  #Encoding the independent variable or feature matrix
# X = np.array(ct.fit_transform(X))
```

```python

le = LabelEncoder()
Y=le.fit_transform(Y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

### Feature Scaling the model
Feature Scaling is applied only on the non encoded values

```python
value_temp=X_train[399]
print(value_temp.reshape(-1,1))
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train[:,:4] = sc.fit_transform(X_train[:,:4])
X_test[:,:4] = sc.transform(X_test[:,:4])
```

```python
value_temp=X_train[399]
print(value_temp.reshape(-1,1))
```

### Applying the RandomForest Classification

```python
from sklearn.ensemble import RandomForestClassifier
classifier_rf = RandomForestClassifier(n_estimators=100,criterion='entropy')
classifier_rf.fit(X_train,Y_train)
```

```python
Y_pred_random_forest = classifier_rf.predict(X_test)
print(np.concatenate((Y_test.reshape(len(Y_test),1),Y_pred_random_forest.reshape(len(Y_pred_random_forest),1)),1))
```

```python

from sklearn.metrics import confusion_matrix, accuracy_score

confusion_matrix_random_forest = confusion_matrix(Y_test,Y_pred_random_forest)
score_random_forest = accuracy_score(Y_test,Y_pred_random_forest)

print(f"The Confusion matix of the Dataset using the Random Forest Model is:\n")
print(confusion_matrix_random_forest)
print(f"The accuracy score of the Dataset using Random Forest Model is : {score_random_forest}")
```

## K Nearest Neighbour
In this classification model
Step 1- choose the number K of how many neighbours we consider.
Step 2 - choose the K nearest neighbour according to the Eucledian Distance between the observing point and the other points\
Step 3 - predict the result from it

```python
from sklearn.neighbors import KNeighborsClassifier
classifier_knn = KNeighborsClassifier(n_neighbors=3,weights='uniform',algorithm='auto',p=3,metric='minkowski')
classifier_knn.fit(X_train,Y_train)
```

```python
Y_pred_knn = classifier_knn.predict(X_test)
print(np.concatenate((Y_test.reshape(len(Y_test),1),Y_pred_knn.reshape(len(Y_pred_knn),1)),1))
```

```python
confusion_matrix_knn = confusion_matrix(Y_test,Y_pred_knn)
score_knn = accuracy_score(Y_test,Y_pred_knn)

print(f"The Confusion matix of the Dataset using the K Nearest Neighbour is:\n")
print(confusion_matrix_knn)
print(f"The accuracy score of the Dataset using K Nearest Neighbour Model is : {score_knn}")
```

# Support Vector Machines

This is an ML algorithm which works by compairing the most extreme parameters which helps in discriminating two different things. SVMs find a hyperplane (a line in 2D or a plane in higher dimensions) that best separates data points belonging to different classes, maximizing the margin (distance) between the hyperplane and the nearest data points from each class. These nearest points are called support vectors.
For Example - If we want to differentiate between apples and oranges, most of the ordinary ml algo will look for the parameters of most applish apple and orangish orange and learn from them. But if we see about the SVMs then it will look for most orangish apple and applish oranges and make the support vectors and hyperplane based on this it becomes very powerful.

## Linear SVM Implemenation

```python
from sklearn.svm import SVC
classifier_svm_linear = SVC(kernel='linear', random_state=0)
classifier_svm_linear.fit(X_train,Y_train)
```

```python
Y_pred_svm_linear = classifier_svm_linear.predict(X_test)
#print(Y_pred_svm_linear)
print(np.concatenate((Y_test.reshape(len(Y_test),1),Y_pred_svm_linear.reshape(len(Y_pred_svm_linear),1)),1))
```

```python
confusion_matrix_svm = confusion_matrix(Y_test,Y_pred_svm_linear)
score_svm = accuracy_score(Y_test,Y_pred_svm_linear)

print(f"The Confusion matix of the Dataset using the SVM Model With Linear Kernal is:\n")
print(confusion_matrix_svm)
print(f"The accuracy score of the Dataset using SVM Model With Linear Kernal is : {score_svm}")
```

## Limitations of SVM With Linear Kernal
In SVM with Linear Kernal , we have straight line which is used to differentate the two zones if we comapare about classification of 2 objects. But in reality most of the time the seperation demands the curved lines rather than seperate so we use a non linear kernal to seperate them. In this way our predicition can improve.

## SVM With Non Linear Kernal

```python
from sklearn.svm import SVC
classifier_svm_kernel = SVC(kernel='rbf', gamma='auto',decision_function_shape='ovr')
classifier_svm_kernel.fit(X_train,Y_train)
```

```python
Y_pred_svm_kernel = classifier_svm_kernel.predict(X_test)
print(np.concatenate((Y_test.reshape(len(Y_test),1),Y_pred_svm_kernel.reshape(len(Y_pred_svm_kernel),1)),1))
```

```python
confusion_matrix_svm = confusion_matrix(Y_test,Y_pred_svm_kernel)
score_svm = accuracy_score(Y_test,Y_pred_svm_kernel)

print(f"The Confusion matix of the Dataset using the SVM Model With Gaussian Kernal is:\n")
print(confusion_matrix_svm)
print(f"The accuracy score of the Dataset using SVM Model With Non Linear Kernal is : {score_svm}")
```
